```python
import pandas
from statistics import mean
```

```python
sizes = {
    "small-19": 5,
    "small-17": 41,
    "medium-25": 85,
    "medium-15": 1.5 * 1024
}
times = {
    "A": {
        "lowest": {
            "small-19": 0.641 * 2,
            "small-17": 2.277 * 2,
            "medium-25": 4.654 * 2,
            "medium-15": 65.581 * 2
        },
        "highest": {
            "small-19": 0.658 * 2,
            "small-17": 3.185 * 2,
            "medium-25": 5.527 * 2,
            "medium-15": 61.256 * 2
        }
    },
    "B": {
        "lowest": {
            "small-19": 0.564 * 2,
            "small-17": 2.197 * 2,
            "medium-25": 4.266 * 2,
            "medium-15": 57.621 * 2
        },
        "highest": {
            "small-19": 0.741 * 2,
            "small-17": 4.87 * 2,
            "medium-25": 8.455 * 2,
            "medium-15": 88.342 * 2
        }
    },
    "C": {
        "lowest": {
            "small-19": 0.639 * 2,
            "small-17": 2.988 * 2,
            "medium-25": 7.381 * 2,
            "medium-15": 105.14 * 2
        },
        "highest": {
            "small-19": 0.649 * 2,
            "small-17": 2.039 * 2,
            "medium-25": 6.973 * 2,
            "medium-15": 95.616 * 2
        }
    }
}
```

```python
instances = 40

discrete_combinations = {
    "A": 4,
    "B": 144,
    "C": 3456
}

table_data = [
    {
        "Algorithm": alg,
        "Total combinations with only discrete variables": discrete_combinations[alg],
    }
    for alg in times.keys()
]

display(pandas.DataFrame(table_data).set_index("Algorithm"))
```

```python
# Possible combinations for each algorithm
capacity_options = 5
reservation_options = 5
cont_options = capacity_options * reservation_options
possible_combinations = {
    "A": cont_options * discrete_combinations["A"],
    "B": cont_options * discrete_combinations["B"],
    "C": cont_options * discrete_combinations["C"]
}

# Keys for size groups
size_groups = {
    "small": ["small-19", "small-17"],
    "medium": ["medium-25", "medium-15"]
}

# Calculate average total times
avg_times = {}
for alg, combos in possible_combinations.items():
    avg_times[alg] = {}
    for group, keys in size_groups.items():
        lowest = mean([min(times[alg]["lowest"][k], times[alg]["highest"][k]) for k in keys])
        highest = mean([max(times[alg]["lowest"][k], times[alg]["highest"][k]) for k in keys])
        avg = (lowest + highest) / 2
        avg_times[alg][group] = avg * instances * combos

# Prepare table data
table_data = [
    {
        "Algorithm": alg,
        "Size Group": group,
        "Estimated Time (hours)": round(avg_times[alg][group] / 3600, 3) # hours
    }
    for alg in possible_combinations.keys() for group in size_groups.keys()
]

total_avg_time = sum([row["Estimated Time (hours)"] for row in table_data])
display(f"Estimated of total time (hours): {total_avg_time:.2e}")

df = pandas.DataFrame(table_data)

display(df.set_index("Algorithm"))
```
