```python
import json
import pandas as pd
```

```python
# Load the JSON file
with open('simulated/results.json', 'r') as f:
    results = json.load(f)

# Flatten the JSON into a DataFrame
rows = []
for entry in results:
    algorithm = entry.get('algorithm')
    instance = entry.get('instance')
    output_mode = entry.get('outputMode')
    executions = entry.get('executions', [])
    for exec in executions:
        row = {
            'algorithm': algorithm,
            'instance': instance,
            'error': exec.get('error', None),
            # 'errorMessage': exec.get('errorMessage', None),
            'value': exec.get('value', None),
            'time': exec.get('time', None)
        }
        # Add parameters if present
        params = exec.get('parameters', {})
        for k, v in params.items():
            row[k] = v
        rows.append(row)

df = pd.DataFrame(rows)
df['maxReservation'] = df['maxReservation'].astype('Int64')
df = df[df['error'] == False]
df = df.drop(columns=['error'])
df['value'] = df['value'] / 3600
df = df.sort_values(by='value', ascending=True)
display(df)
```
