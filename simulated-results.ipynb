```python
import json
import pandas as pd
```

```python
# Load the JSON file
with open('simulated/results.json', 'r') as f:
    results = json.load(f)

# Flatten the JSON into a DataFrame
rows = []
for entry in results:
    algorithm = entry.get('algorithm')
    instance = entry.get('instance')
    output_mode = entry.get('outputMode')
    executions = entry.get('executions', [])
    for exec in executions:
        row = {
            'algorithm': algorithm,
            'instance': instance,
            'value': exec.get('value', None),
            'time': exec.get('time', None),
            'error': exec.get('error', None),
            #'errorMessage': exec.get('errorMessage', None),
        }
        # Add parameters if present
        params = exec.get('parameters', {})
        for k, v in params.items():
            row[k] = v
        rows.append(row)

df = pd.DataFrame(rows)
df['maxReservation'] = df['maxReservation'].astype('Int64')
df = df[df['error'] == False]
df = df.drop(columns=['error'])
df['value'] = df['value'] / 3600
df = df.sort_values(by='value', ascending=True)
display(df)
```

```python
capacity = 1000
instance_type = 'small'
df = df[df['instance'].str.contains(instance_type)]
df_c = df[df['capacity'] == capacity]
df_c = df_c.drop(columns=['capacity'])
display(df_c)
```

```python
def show_best_times(df):
    best_times = df.groupby('instance').apply(lambda x: x.loc[x['value'].idxmin()]).reset_index(drop=True)
    return best_times

best_algorithms = show_best_times(df)
display(best_algorithms)
```

```python
best_times_b = show_best_times(df[df['algorithm'] == 'B'])
display(best_times_b)
```

```python
best_times_a = show_best_times(df[df['algorithm'] == 'A'])
display(best_times_a)
```

```python
capacity = 50
df_capacity = df[df['capacity'] == capacity]
display(show_best_times(df_capacity))
display(show_best_times(df_capacity[df_capacity['algorithm'] == 'C']))
display(show_best_times(df_capacity[df_capacity['algorithm'] == 'B']))
display(show_best_times(df_capacity[df_capacity['algorithm'] == 'A']))
```
